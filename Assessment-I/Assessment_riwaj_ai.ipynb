```python
def remove_invalid_countries(df):
    invalid_entries = [
        'Very high human development',
        'High human development', 'Medium human development',
        'Low human development', 'Arab States',
        'East Asia and the Pacific', 'Europe and Central Asia',
        'Latin America and the Caribbean', 'South Asia',
        'Sub-Saharan Africa', 'World'
    ]

    # Strip leading/trailing whitespace from 'country' column
    df['country'] = df['country'].str.strip()

    # Drop rows where 'country' is in invalid_entries
    return df.drop(df[df['country'].isin(invalid_entries)].index)
```

```python
def clean_missing_values(df, threshold=10):
    """
    Cleans missing values in a DataFrame.

    Parameters:
    df (pd.DataFrame): The DataFrame to clean.
    threshold (float): Percentage threshold to decide how to handle missing values.
                       Columns with missing values > threshold% are filled with the mean.
                       Otherwise, rows with missing values are dropped.

    Returns:
    pd.DataFrame: Cleaned DataFrame.
    """

    print("Number of missing values before cleaning:", df.isnull().sum().sum())
    print("Number of rows before cleaning:", df.shape[0])

    # Calculate missing percentage per column
    missing_info = df.isnull().sum() / len(df) * 100

    for column in df.columns:
        if missing_info[column] > threshold:
            # Fill numeric columns with mean
            df[column] = df[column].fillna(df[column].mean())

        else:
            # Drop rows with missing values
            df = df.dropna(subset=[column])

    print("Number of missing values after cleaning:", df.isnull().sum().sum())
    print("Number of rows after cleaning:", df.shape[0])

    return df
```

```python
# Clean the HDI 2020–2022 dataset by handling missing values:
# Columns with greater than 10% missing are filled with the mean (numeric) or mode (non-numeric)
# Columns with ≤10% missing have rows with missing values dropped
hdi_2022_df = clean_missing_values(hdi_2022_df, threshold=10)
```

```python
#Detecting outliers in HDI

#Getting the Q1 of the hdi column
q1_hdi_south_asian = hdi_south_asian["hdi"].quantile(0.25)

#Getting the Q3 of the hdi column
q3_hdi_south_asian = hdi_south_asian["hdi"].quantile(0.75)

#Calculating IQR
IQR_hdi = q3_hdi_south_asian - q1_hdi_south_asian

#Calculating upper bound of HDI column
lower_hdi = q1_hdi_south_asian - 1.5 * IQR_hdi

#Calculating lower bound of HDI column
upper_hdi = q3_hdi_south_asian + 1.5 * IQR_hdi
```

```python
#Detecting outliers in GNI per capita

#Getting the Q1 of the GNI column
q1_gni_south_asian = hdi_south_asian["gross_inc_percap"].quantile(0.25)

#Getting the Q3 of the GNI column
q3_gni_south_asian = hdi_south_asian["gross_inc_percap"].quantile(0.75)

#Calculating IQR
IQR_gni = q3_gni_south_asian - q1_gni_south_asian

#Calculating upper bound of GNI column
lower_gni = q1_gni_south_asian - 1.5 * IQR_gni

#Calculating lower bound of GNI column
upper_gni = q3_gni_south_asian + 1.5 * IQR_gni
```

```python
hdi_south_asian["hdi_outlier"] = (
    (hdi_south_asian["hdi"] < lower_hdi) |
    (hdi_south_asian["hdi"] > upper_hdi)
)

# Show the outlier countries
hdi_south_asian[hdi_south_asian["hdi_outlier"]][["country", "hdi"]]
```

```python
# Mark countries as GNI outliers
hdi_south_asian["gni_outlier"] = (
    (hdi_south_asian["gross_inc_percap"] < lower_gni) |
    (hdi_south_asian["gross_inc_percap"] > upper_gni)
)

# See the outlier countries
gni_outliers = hdi_south_asian[hdi_south_asian["gni_outlier"]][["country", "gross_inc_percap"]]
gni_outliers
```

```python
hdi_south_asian["outlier"] = (
    hdi_south_asian["hdi_outlier"] | hdi_south_asian["gni_outlier"]
)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))

# Plot normal points
plt.scatter(
    hdi_south_asian.loc[~hdi_south_asian["outlier"], "gross_inc_percap"],
    hdi_south_asian.loc[~hdi_south_asian["outlier"], "hdi"],
    color="blue",
    label="Normal",
    alpha=0.7,
    s=80
)

# Plot outliers
plt.scatter(
    hdi_south_asian.loc[hdi_south_asian["outlier"], "gross_inc_percap"],
    hdi_south_asian.loc[hdi_south_asian["outlier"], "hdi"],
    color="red",
    label="Outlier",
    alpha=0.9,
    s=120
)

plt.xlabel("GNI per Capita (USD)")
plt.ylabel("HDI")
plt.title("GNI per Capita vs HDI (Outliers Highlighted)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Function to plot scatter + trendline
def scatter_trendline(x, y, ax, title, color, corr_value):
    ax.scatter(x, y, color=color)
    # Trendline
    m, b = np.polyfit(x, y, 1)
    ax.plot(x, m*x + b, color="red")
    # Labels and title
    ax.set_xlabel(title.split(": ")[1])
    ax.set_ylabel("HDI")
    ax.set_title(title)
    # Annotate correlation
    ax.text(0.05, 0.95, f"r = {round(corr_value,2)}", transform=ax.transAxes,
            fontsize=12, verticalalignment='top', bbox=dict(facecolor='white', alpha=0.5))

# Create 2x2 plot
fig, axes = plt.subplots(2, 2, figsize=(14,10))

# South Asia: Gender Development
scatter_trendline(
    hdi_prob1b_df_south_asian["gender_development"],
    hdi_prob1b_df_south_asian["hdi"],
    axes[0,0],
    "South Asia: Gender Development",
    "blue",
    corr_south_gender
)

# South Asia: Life Expectancy
scatter_trendline(
    hdi_prob1b_df_south_asian["life_expectancy"],
    hdi_prob1b_df_south_asian["hdi"],
    axes[0,1],
    "South Asia: Life Expectancy",
    "blue",
    corr_south_life
)

# Middle East: Gender Development
scatter_trendline(
    hdi_prob1b_df_middle_east_countries["gender_development"],
    hdi_prob1b_df_middle_east_countries["hdi"],
    axes[1,0],
    "Middle East: Gender Development",
    "green",
    corr_middle_gender
)

# Middle East: Life Expectancy
scatter_trendline(
    hdi_prob1b_df_middle_east_countries["life_expectancy"],
    hdi_prob1b_df_middle_east_countries["hdi"],
    axes[1,1],
    "Middle East: Life Expectancy",
    "green",
    corr_middle_life
)

plt.tight_layout()
plt.show()
```

In both regions, countries with higher life expectancy and better gender development tend to have higher HDI. In South Asia, this pattern exists but is not very strong, showing quite a bit of variation, which suggests that other factors also play a role in determining HDI. In the Middle East, the link is much clearer and more consistent, especially when it comes to gender development. Overall, life expectancy and gender equality both influence human development, but their effect seems stronger and more reliable in the Middle East than in South Asia.

```python
# Function to detect outliers using 1.5 * IQR rule
def get_outliers(series):
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    # Return indices of outliers
    return series[(series < lower) | (series > upper)].index
```

```python
#Outliers for South Asia
outlier_hdi_south = get_outliers(hdi_prob1b_df_south_asian["hdi"])
outlier_gni_south = get_outliers(hdi_prob1b_df_south_asian["gross_inc_percap"])

print("South Asia HDI outliers:")
print(hdi_prob1b_df_south_asian.loc[outlier_hdi_south, ["country", "hdi"]])
print("\nSouth Asia GNI per Capita outliers:")
print(hdi_prob1b_df_south_asian.loc[outlier_gni_south, ["country", "gross_inc_percap"]])
```
